```python
import openml
import os
```

```python
# Define the suite (e.g., '269' is the AutoML Benchmark suite)
suite = openml.study.get_suite(269)
output_dir = "./automl_datasets"
os.makedirs(output_dir, exist_ok=True)

for data_id in suite.data:
    dataset = openml.datasets.get_dataset(data_id)
    # Get the data as a dataframe
    X, y, _, _ = dataset.get_data(dataset_format="dataframe")
    
    # Save to CSV
    filename = f"{dataset.name}.csv"
    X.to_csv(os.path.join(output_dir, filename), index=False)
    print(f"Downloaded: {filename}")
```
