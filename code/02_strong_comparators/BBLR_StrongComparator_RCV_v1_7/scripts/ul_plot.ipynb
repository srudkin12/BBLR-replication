```python
dgp = 'UL'
test_flag = 'test'  # 'test' or 'train'
results_folder_name = 'UL'
metric = 'B1_rmse'


csv_files_names = [f'{dgp}_BBLR_0.0_1000_normal_up_to_N_OLS_mean_{test_flag}.csv',
    f'{dgp}_LOWESS_0.0_1000_normal_{test_flag}.csv',
    f'{dgp}_LR_0.0_1000_normal_{test_flag}.csv',
    f'{dgp}_LPR2_0.0_1000_normal_{test_flag}.csv']
csv_files_paths = [f'cluster_results/{results_folder_name}/{name}' for name in csv_files_names]

a_vectors_file_path = f'cluster_results/{results_folder_name}/{dgp}_a_vectors.csv'
r2_vectors_file_path = f'cluster_results/{results_folder_name}/{dgp}_r2_scores.csv'

a_df = pd.read_csv(a_vectors_file_path)
# abs_sum_a2_a3 = np.abs(a_df['a2']) + np.abs(a_df['a3'])
# abs_sum_a2_a3 = np.abs(a_df['a2'] + a_df['a3'])

results_dfs = []
for csv_file_path in csv_files_paths:
    df = pd.read_csv(csv_file_path)
    results_dfs.append(df[metric])

methods = ['BBLR', 'LOWESS', 'LR', 'LPR2']

df_all = pd.DataFrame({
    f'{metric}_{methods[i]}':results_dfs[i] for i in range(len(methods))
    })

df_all['a'] = a_df['a2']

plt.scatter(df_all['a'], df_all[f'{metric}_BBLR'], label='BBLR', s=0.5)
plt.scatter(df_all['a'], df_all[f'{metric}_LOWESS'], label='LOWESS', s=0.5)
plt.scatter(df_all['a'], df_all[f'{metric}_LR'], label='LR', s=0.5)
plt.scatter(df_all['a'], df_all[f'{metric}_LPR2'], label='LPR2', s=0.5)
plt.xlabel('a (slope)')
plt.ylabel(metric)
plt.legend()
plt.savefig(f'figures/{dgp}_{metric}_vs_a_without.pdf')
```
