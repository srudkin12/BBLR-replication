```python
i = 42

# Generate Friedman1 dataset with 10 variables and seed=42
X, y = make_friedman1(n_samples=1000, n_features=10, noise=1.0, random_state=i)
df = pd.DataFrame(X)
df['target'] = y

X = df.drop('target', axis=1)
y = df['target']

# Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=i+10000)

X_train = np.array(X_train)
y_train = np.array(y_train)
X_test = np.array(X_test)
y_test = np.array(y_test)

column_names = [f'x{i+1}' for i in range(X_train.shape[1])]

scaler = StandardScaler()
X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train), columns=column_names)
X_test_scaled = pd.DataFrame(scaler.transform(X_test), columns=column_names)

X_train_scaled = np.array(X_train_scaled)
y_train = np.array(y_train)
X_test_scaled = np.array(X_test_scaled)
y_test = np.array(y_test)
```

```python
def run(
    M=5,
    n=1000,
    dgp_name="BF",
    param=0.7,
    method="BBLR",
    d=2,
    noise_var=0.2,
    x_distribution='normal',
    min_pts=10,
    eps_range=(0.1,3.0),
    eps_multiplier=0.1,
    bblr_outliers_method="up_to_N",
    coef_comp='mean',
    bblr_regression_type="OLS",
    bblr_outliers_method_pred="up_to_N",
    test_size=0.5,
    include_training_data=False
    ):

# def monte_carlo_estimates(M, n, dgp_name, param, method, d, noise_var=0.2, min_pts=10, eps_range=(0.1,1.5), x_distribution='normal'):
    def one_iteration(i):
        # print(i)
        df = generate_data(n, dgp_name, d=d, noise_var=noise_var, seed=i, x_distribution=x_distribution)
        # define column names
        columns_list = [f'X{i+1}' for i in range(d)]
        X = df[columns_list]
        y = df['y']
        X = X.to_numpy()
        y = y.to_numpy()
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=42)

        result_df = estimate_coefs(X_train, X_test, y_train, y_test, method,
                                    n=n, dgp_name=dgp_name, param=param, column_names=columns_list,  
                                    bblr_regression_type = bblr_regression_type, eps_multiplier = eps_multiplier,
                                    coef_comp = coef_comp, ci=False, pred_outliers_method=bblr_outliers_method_pred)
        results_rows = aggregate_raw_results(X_train, X_test, result_df, dgp_name, method, i, 'test')

        if include_training_data:
            # we set the optimal parameter to avoid repeating the process of searching for optimal parameter
            if param == 0:
                training_param = results_rows['parameter']
            else:
                training_param = param
            result_df_train = estimate_coefs(X_train, X_train, y_train, y_train, method,
                                        n=n, dgp_name=dgp_name, param=training_param, column_names=columns_list,  
                                        bblr_regression_type = bblr_regression_type, eps_multiplier = eps_multiplier,
                                        coef_comp = coef_comp, ci=False, pred_outliers_method=bblr_outliers_method_pred)
            results_rows_train = aggregate_raw_results(X_train, X_train, result_df_train, dgp_name, method, i, 'train')
        else:
            results_rows_train = None

        return results_rows, results_rows_train

    # for debugging
    for iter in range(M):
        one_iteration(iter)

    records = Parallel(n_jobs=-1)(delayed(one_iteration)(iter) for iter in range(M))

    results, results_train = zip(*records)

    results = pd.DataFrame(results)
    results_train = pd.DataFrame(results_train)

    save_aggregated_results(results, n, method, dgp_name, x_distribution, 'test', param=param,
                            bblr_outliers_method=bblr_outliers_method,
                            coef_comp=coef_comp,
                            bblr_regression_type=bblr_regression_type)
    if include_training_data:
        save_aggregated_results(results_train, n, method, dgp_name, x_distribution, 'train', param=param,
                            bblr_outliers_method=bblr_outliers_method,
                            coef_comp=coef_comp,
                            bblr_regression_type=bblr_regression_type)

    return results, results_train
```

```python
# raw results:
n=1000
dgp_name="BB"
# param=0.7
# method="BBLR"
param=0.3
method="LOWESS"
d=2
noise_var=0.2
x_distribution='normal'
min_pts=10
eps_range=(0.1,3.0)
eps_multiplier=0.1
bblr_outliers_method="up_to_N"
coef_comp='mean'
bblr_regression_type="OLS"
bblr_outliers_method_pred="up_to_N"
test_size=0.5
include_training_data=False
ci=True
confidence=0.95


random_seed = 42
df = generate_data(n, dgp_name, d=d, noise_var=noise_var, seed=random_seed, x_distribution=x_distribution)
# define column names
columns_list = [f'X{i+1}' for i in range(d)]
X = df[columns_list]
y = df['y']
X = X.to_numpy()
y = y.to_numpy()
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=42)

params = [0.4, 0.5, 0.6]  # no i etcetera

for param in params:
    results_df = estimate_coefs(X_train, X_test, y_train, y_test, method,
                                n=n, dgp_name=dgp_name, param=param, column_names=columns_list, min_points=min_pts,  
                                bblr_regression_type = bblr_regression_type, eps_multiplier = eps_multiplier,
                                coef_comp = coef_comp, ci=ci, pred_outliers_method=bblr_outliers_method_pred,
                                confidence = confidence, bblr_outliers_method=bblr_outliers_method,
                                eps_range = eps_range)
    save_raw_results(results_df, n, method, dgp_name, x_distribution, 'test',
                        bblr_outliers_method=bblr_outliers_method,
                        coef_comp=coef_comp,
                        bblr_regression_type=bblr_regression_type, 
                        ci=ci)

    results_df = estimate_coefs(X_train, X_train, y_train, y_train, method,
                                n=n, dgp_name=dgp_name, param=param, column_names=columns_list, min_points=min_pts,  
                                bblr_regression_type = bblr_regression_type, eps_multiplier = eps_multiplier,
                                coef_comp = coef_comp, ci=ci, pred_outliers_method=bblr_outliers_method_pred,
                                confidence = confidence, bblr_outliers_method=bblr_outliers_method,
                                eps_range = eps_range)
    save_raw_results(results_df, n, method, dgp_name, x_distribution, 'train',
                        bblr_outliers_method=bblr_outliers_method,
                        coef_comp=coef_comp,
                        bblr_regression_type=bblr_regression_type, 
                        ci=ci)
```
