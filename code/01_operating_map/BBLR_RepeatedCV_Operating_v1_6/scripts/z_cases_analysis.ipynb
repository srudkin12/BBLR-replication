```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
mean_beta_rmse_all = []
for dgp in ['Z1', 'Z2', 'Z3', 'Z4', 'Z5']:
    csv_files_names = [f'{dgp}_BBLR_0.0_1000_normal_up_to_N_OLS_mean_test.csv',
            f'{dgp}_LOWESS_0.0_1000_normal_test.csv',
            f'{dgp}_LPR2_0.0_1000_normal_test.csv']
    mean_beta_rmse_case = []
    for csv_file_name in csv_files_names:
        df = pd.read_csv(f'cluster_results/M1000sigma05/{csv_file_name}')
        B_rmse_cols = [col for col in df.columns if col.startswith('B') and col.endswith('_rmse')]
        B_rmse_cols.remove('B0_rmse')  # Remove B0_rmse from the list
        B_rmse_cols_means = df[B_rmse_cols].mean().tolist()  # Get the mean of each B_rmse column
        mean_B_rmse_cols_means = np.mean(B_rmse_cols_means)  # Get the mean of the means
        mean_beta_rmse_case.append(mean_B_rmse_cols_means)  # Append the mean
    mean_beta_rmse_all.append(mean_beta_rmse_case)  # Append the list for this case
```

```python
mean_beta_rmse_all
```

```python
mean_beta_rmse_all_df = pd.DataFrame(mean_beta_rmse_all, columns=['BBLR', 'LOWESS', 'LPR2'], index=['Z1', 'Z2', 'Z3', 'Z4', 'Z5'])
mean_beta_rmse_all_df
```

```python
mean_y_rmse_all = []
for dgp in ['Z1', 'Z2', 'Z3', 'Z4', 'Z5']:
    csv_files_names = [f'{dgp}_BBLR_0.0_1000_normal_up_to_N_OLS_mean_test.csv',
            f'{dgp}_LOWESS_0.0_1000_normal_test.csv',
            f'{dgp}_LPR2_0.0_1000_normal_test.csv']
    mean_y_rmse_case = []
    for csv_file_name in csv_files_names:
        df = pd.read_csv(f'cluster_results/M1000sigma05/{csv_file_name}')
        y_rmse_cols_means = df['y_rmse'].mean() # Get the mean of each B_rmse column
        mean_y_rmse_case.append(y_rmse_cols_means)  # Append the mean
    mean_y_rmse_all.append(mean_y_rmse_case)  # Append the list for this case
```

```python
mean_y_rmse_all_df = pd.DataFrame(mean_y_rmse_all, columns=['BBLR', 'LOWESS', 'LPR2'], index=['Z1', 'Z2', 'Z3', 'Z4', 'Z5'])
mean_y_rmse_all_df
```

```python
# concat mean_beta_rmse_all_df and mean_y_rmse_all_df
combined_df = pd.concat([mean_beta_rmse_all_df, mean_y_rmse_all_df], axis=1, keys=['Mean Beta RMSE', 'Mean Y RMSE'])
combined_df
```

```python
# convert combined_df to LaTeX table and save to txt file
latex_table = combined_df.to_latex(float_format="%.3f", caption="Mean RMSE for Beta and Y across different DGPs and methods", label="tab:mean_rmse")
with open("tables/z_table.txt", "w") as f:
    f.write(latex_table) 
```
